```python
# Clean text
def clean_text(text):
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'([a-z])([A-Z])', r'\1 \2', text)
    text = re.sub(r'[^a-zA-Z0-9.,?! ]', '', text)
    return text.strip()
```

```python
# Process all transcript files
for file in os.listdir(input_folder):
    if file.endswith(".txt"):

        input_path = os.path.join(input_folder, file)
        output_path = os.path.join(output_folder, file)

        # Read file
        with open(input_path, "r", encoding="utf-8") as f:
            text = f.read()

        # Clean text 
        cleaned_text = clean_text(text)

       # save processed file
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(cleaned_text)

        print(f"Processed:{file}")
```

```python
# Split text into individual sentences (each sentence = one subtitle)

def split_into_sentences(text):
    sentences = re.split(r'[.?!]', text)

    # clean sentences
    sentences = [s.strip() for s in sentences if s.strip()]

    return sentences
```

```python
# check sentence
sentences = split_into_sentences(text)

for i, line in enumerate(sentences, 1):
    print(f"{i}: {line}")
```

```python
# We break long sentences into smaller lines 
def split_for_subtitles(text, max_words=8):
    words = text.split()
    lines = []

    for i in range(0, len(words), max_words):
       line = " ".join(words[i:i+max_words])
       lines.append(line)

    return lines
```

```python
def generate_subtitle_lines(text):
    final_lines = []
    sentences = re.split(r'[.?!]', text)

    for sentence in sentences:
        sentence = sentence.strip()
        if sentence:
            small_lines = split_for_subtitles(sentence)
            final_lines.extend(small_lines)
    return final_lines            
```

```python
# Assign time duration for each sentence (e.g., 2–3 seconds)

def format_time(seconds):
    hrs = int(seconds // 3600)
    mins = int((seconds % 3600) // 60)
    secs = int(seconds % 60)
    return f"{hrs:02}:{mins:02}:{secs:02},000"
```

```python
subtitle_folder = r"D:/Project/Subtitle Generator and Summarizer/Subtitles"
os.makedirs(subtitle_folder, exist_ok=True)

for file in os.listdir(input_folder):
    if file.endswith(".txt"):
        file_path = os.path.join(input_folder, file)
        with open(file_path, "r", encoding="utf-8") as f:
            text = f.read()

        subtitle_lines = generate_subtitle_lines(text)

        # Save one .srt file per audio
        srt_name = file.replace(".txt", ".srt")
        output_path = os.path.join(subtitle_folder, srt_name)

        with open(output_path, "w", encoding="utf-8") as f:
            time = 0
            for i, line in enumerate(subtitle_lines, 1):
                start = format_time(time)
                time += 3
                end = format_time(time)
                f.write(f"{i}\n{start} --> {end}\n{line}\n\n")

        print(f"Subtitle saved: {srt_name}")
```

# **Text Summarization**

```python
input_folder = r"D:/Project/Subtitle Generator and Summarizer/Processed_Transcripts"
output_folder = r"D:/Project/Subtitle Generator and Summarizer/Summaries"

os.makedirs(output_folder, exist_ok=True)
```

```python
# Apply NLP models (BART)
bart_summarizer = pipeline("summarization", model="facebook/bart-large-cnn" ) 
```

```python
# Generate concise summaries (3–5 sentences + ~100 words)

input_folder = r"D:/Project/Subtitle Generator and Summarizer/Processed_Transcripts"
output_folder = r"D:/Project/Subtitle Generator and Summarizer/Summaries"
os.makedirs(output_folder, exist_ok=True)

def chunk_text(text, max_words=200):
    words = text.split()
    chunks = []
    for i in range(0, len(words), max_words):
        chunk = " ".join(words[i:i+max_words])
        chunks.append(chunk)
    return chunks

for file in os.listdir(input_folder):
    if file.endswith(".txt"):

        input_path = os.path.join(input_folder, file)
        output_path = os.path.join(output_folder, file)

        # Read input text
        with open(input_path, "r", encoding="utf-8") as f:
            text = f.read()

        # Split into chunks 
        chunks = chunk_text(text, max_words=200)

        all_summaries = []

        for chunk in chunks:
            prompt = "Summarize the following text clearly:\n" + chunk

            # Safe length for BART
            summary = bart_summarizer(
                prompt,
                max_length=130,
                min_length=50,
                do_sample=False
            )[0]['summary_text']

            all_summaries.append(summary)

        # Combine all chunk summaries
        final_summary = " ".join(all_summaries)

        # Limit to 140 words
        words = final_summary.split()
        if len(words) > 140:
            final_summary = " ".join(words[:140])

        # Save summary
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(final_summary)

        print(f"Summary saved for: {file}")
```
